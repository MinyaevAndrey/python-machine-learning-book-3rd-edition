### Applying mathematical operations to tensors

```python
tf.random.set_seed(1)

t1 = tf.random.uniform(shape=(5, 2),
                       minval=-1.0,
                       maxval=1.0)

t2 = tf.random.normal(shape=(5, 2),
                      mean=0.0,
                      stddev=1.0)
```

```python
t1
```

```python
#
# поэлементное произведение
#
t3 = tf.multiply(t1, t2).numpy()
print(t3)
```

```python
#
# среднее, сумма и стандартное отклонение по определенной оси
#
t4 = tf.math.reduce_mean(t1, axis=0) # по столбцам
print(t4)
# tf.math.reduce_mean
# tf.math.reduce_sun
# tf.math.reduce_std
```

```python
#
# матричное произведение, указать кого транспонировать
#
# Матричное умножение требует: (5, 2) × (2, 5) или (5, 2) × (2, n)
# Но у нас (5, 2) × (5, 2) - внутренние размеры 2 и 5 НЕ СОВПАДАЮТ!
t5 = tf.linalg.matmul(t1, t2, transpose_b=True)
print(t5.numpy())
```

```python
#
# матричное произведение
#
t6 = tf.linalg.matmul(t1, t2, transpose_a=True)
print(t6.numpy())
```

```python
#
# матричное произведение
#
A = tf.constant([[1, 2],    # форма (2, 2)
                 [3, 4]])
B = tf.constant([[5, 6],    # форма (2, 2)
                 [7, 8]])

# Обычное умножение: (2,2) × (2,2) = (2,2)
C = tf.matmul(A, B)
print(C.numpy())
```

### Split, stack, and concatenate tensors

```python
#
# разделяет входной тензор на список тензоров равных размеров.
#
tf.random.set_seed(1)
t = tf.random.uniform((6,))
print(t.numpy())

t_splits = tf.split(t, 3)
[item.numpy() for item in t_splits]
```

```python
#
# разделяет входной тензор по указанным размерами
#
tf.random.set_seed(1)
t = tf.random.uniform((5,))
print(t.numpy())

t_splits = tf.split(t, num_or_size_splits=[3, 2])
[item.numpy() for item in t_splits]
```

```python
#
# Объединение
#
A = tf.ones((3,))
B = tf.zeros((2,))
C = tf.concat([A, B], axis=0)
print(C.numpy())
```

```python
A = tf.ones((3,))
B = tf.zeros((3,))
S = tf.stack([A, B], axis=1)
print(S.numpy())
```

### Combining two tensors into a joint dataset

```python
tf.random.set_seed(1)

t_x = tf.random.uniform([4, 3], dtype=tf.float32)
t_y = tf.range(4)
```

```python
ds_x = tf.data.Dataset.from_tensor_slices(t_x)
ds_y = tf.data.Dataset.from_tensor_slices(t_y)

ds_joint = tf.data.Dataset.zip((ds_x, ds_y))

for example in ds_joint:
    print('  x: ', example[0].numpy(),
          '  y: ', example[1].numpy())
```

```python
## method 2:
ds_joint = tf.data.Dataset.from_tensor_slices((t_x, t_y))

for example in ds_joint:
    print('  x: ', example[0].numpy(),
          '  y: ', example[1].numpy())
```

```python
ds_trans = ds_joint.map(lambda x, y: (x*2-1.0, y))

for example in ds_trans:
    print('  x: ', example[0].numpy(),
          '  y: ', example[1].numpy())
```

### Shuffle, batch, and repeat

```python
tf.random.set_seed(1)
ds = ds_joint.shuffle(buffer_size=len(t_x))

for example in ds:
    print('  x: ', example[0].numpy(),
          '  y: ', example[1].numpy())
```

```python
ds = ds_joint.batch(batch_size=3,
                    drop_remainder=False)

batch_x, batch_y = next(iter(ds))

print('Batch-x: \n', batch_x.numpy())

print('Batch-y:   ', batch_y.numpy())
```

```python
ds = ds_joint.batch(3).repeat(count=2)

for i,(batch_x, batch_y) in enumerate(ds):
    print(i, batch_x.shape, batch_y.numpy())
```

```python
ds = ds_joint.repeat(count=2).batch(3)
for i,(batch_x, batch_y) in enumerate(ds):
    print(i, batch_x.numpy(), batch_y.numpy())
```

```python
tf.random.set_seed(1)

## Order 1: shuffle -> batch -> repeat
ds = ds_joint.shuffle(4).batch(2).repeat(3)

for i,(batch_x, batch_y) in enumerate(ds):
    print(i, batch_x.shape, batch_y.numpy())
```

```python
tf.random.set_seed(1)

## Order 1: shuffle -> batch -> repeat
ds = ds_joint.shuffle(4).batch(2).repeat(20)

for i,(batch_x, batch_y) in enumerate(ds):
    print(i, batch_x.shape, batch_y.numpy())
```

```python
tf.random.set_seed(1)

## Order 2: batch -> shuffle -> repeat
ds = ds_joint.batch(2).shuffle(4).repeat(3)

for i,(batch_x, batch_y) in enumerate(ds):
    print(i, batch_x.shape, batch_y.numpy())
```

```python
tf.random.set_seed(1)

## Order 2: batch -> shuffle -> repeat
ds = ds_joint.batch(2).shuffle(4).repeat(20)

for i,(batch_x, batch_y) in enumerate(ds):
    print(i, batch_x.shape, batch_y.numpy())
```

```python
import pathlib
imgdir_path = pathlib.Path('cat_dog_images')
file_list = sorted([str(path) for path in imgdir_path.glob('*.jpg')])
print(file_list)
```

```python
import matplotlib.pyplot as plt
import os


fig = plt.figure(figsize=(7, 4))
for i,file in enumerate(file_list):
    img_raw = tf.io.read_file(file)
    img = tf.image.decode_image(img_raw)
    print('Image shape: ', img.shape)
    ax = fig.add_subplot(2, 3, i+1)
    ax.set_xticks([]); ax.set_yticks([])
    ax.imshow(img)
    ax.set_title(os.path.basename(file), size=15)

# plt.savefig('ch13-catdot-examples.pdf')
plt.tight_layout()
plt.show()
```

```python
#
# метки из имен файлов
#
labels = [1 if 'dog' in os.path.basename(file) else 0
          for file in file_list]
print(labels)
```

```python
#
# слить имена и метки в один датасет-тензор
#
ds_files_labels = tf.data.Dataset.from_tensor_slices((file_list, labels))

for item in ds_files_labels:
    print(item[0].numpy(), item[1].numpy())
```

```python
#
# закачка содержимого файлов в датасет
#

# функция обработки одного файла
def load_and_preprocess(path, label):
    image = tf.io.read_file(path)
    image = tf.image.decode_jpeg(image, channels=3)
    image = tf.image.resize(image, [img_height, img_width])
    image /= 255.0

    return image, label

img_width, img_height = 120, 80

# обработать все файлы и залить их в новый датасет ds_images_labels
ds_images_labels = ds_files_labels.map(load_and_preprocess)

fig = plt.figure(figsize=(7, 4))
for i,example in enumerate(ds_images_labels):
    print(example[0].shape, example[1].numpy())
    ax = fig.add_subplot(2, 3, i+1)
    ax.set_xticks([]); ax.set_yticks([])
    ax.imshow(example[0])
    ax.set_title('{}'.format(example[1].numpy()),
                 size=15)

plt.tight_layout()
#plt.savefig('ch13-catdog-dataset.pdf')
plt.show()
```

Alternative ways for loading a dataset

```python
mnist, mnist_info = tfds.load('mnist',
                              with_info=True,
                              shuffle_files=False
)
print(mnist_info)
print(mnist.keys())
```

```python
ds_train = mnist['train']

assert isinstance(ds_train, tf.data.Dataset)

# важное преобразование из словаря в кортеж
ds_train = ds_train.map(lambda item: (item['image'], item['label']))

ds_train = ds_train.batch(10)
batch = next(iter(ds_train))
print(batch[0].shape, batch[1])
# batch — это кортеж из двух элементов: (изображения, метки)
# batch[0] — сами изображения (тензор формы (10, 28, 28, 1))
# batch[1] — соответствующие метки (тензор формы (10,))

fig = plt.figure(figsize=(15, 6))
for i,(image,label) in enumerate(zip(batch[0], batch[1])):
    ax = fig.add_subplot(2, 5, i+1)
    ax.set_xticks([]); ax.set_yticks([])
    ax.imshow(image[:, :, 0], cmap='gray_r')
    ax.set_title('{}'.format(label), size=15)

plt.show()
```
